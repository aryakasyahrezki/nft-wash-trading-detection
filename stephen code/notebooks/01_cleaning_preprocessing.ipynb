```python
import sqlite3
import pandas as pd
from pathlib import Path
```

# SETUP FOLDER DAN CEK KONEKSI DATABASE

```python
RAW_DIR = Path("../data/raw")
RAW_DIR.mkdir(parents=True, exist_ok=True)

db_path = RAW_DIR / "nfts.sqlite"

conn = sqlite3.connect(db_path)
```

# QUERY DATA

```python
JUNE_1 = 1622505600
SEPT_1 = 1630454400

query = f"""
SELECT
    t.transaction_hash,
    t.block_number,
    t.timestamp,
    t.nft_address,
    t.token_id,
    t.from_address,
    t.to_address,
    t.transaction_value,
    m.timestamp AS mint_timestamp,
    sf.transfers_out AS transfers_out_from,
    sf.transfers_in AS transfers_in_from,
    st.transfers_out AS transfers_out_to,
    st.transfers_in AS transfers_in_to,
    ot.num_transitions
FROM transfers t
LEFT JOIN mints m
    ON t.token_id = m.token_id
    AND t.nft_address = m.nft_address
LEFT JOIN transfer_statistics_by_address sf
    ON t.from_address = sf.address
LEFT JOIN transfer_statistics_by_address st
    ON t.to_address = st.address
LEFT JOIN ownership_transitions ot
    ON t.from_address = ot.from_address
    AND t.to_address = ot.to_address
WHERE t.timestamp >= {JUNE_1}
  AND t.timestamp < {SEPT_1}
"""
```

# DATAFRAME

```python
df = pd.read_sql_query(query, conn)

conn.close()

print(df.shape)

df.head()
```

# OUTPUT CSV

```python
output_path = RAW_DIR / "nft_transfers_jun_aug.csv"

df.to_csv(output_path, index=False)

print("Saved to:", output_path)
```

# LOAD DATASET

```python
df = pd.read_csv("../data/raw/nft_transfers_jun_aug.csv")

print(df.shape)
df.head()
```

# MISSING VALUE

```python
df.isnull().sum().sort_values(ascending=False)
```

# KONVERSI TIMESTAMP

```python
df["timestamp_dt"] = pd.to_datetime(
    df["timestamp"],
    unit="s"
)

df["mint_timestamp_dt"] = pd.to_datetime(
    df["mint_timestamp"],
    unit="s",
    errors="coerce"
)
```

```python
df["transaction_value"] = pd.to_numeric(
    df["transaction_value"],
    errors="coerce"
)

df["transaction_value"] = df["transaction_value"].fillna(0)
```

```python
df = df[df["transaction_value"] > 0].copy()

print(df.shape)
```

# REMOVE BURN ADDRESS

```python
burn_addresses = [
    "0x0000000000000000000000000000000000000000",
    "0x000000000000000000000000000000000000dead"
]

df = df[
    (~df["from_address"].isin(burn_addresses)) &
    (~df["to_address"].isin(burn_addresses))
].copy()

print(df.shape)
```

# DROP DUPLICATE

```python
df = df.drop_duplicates(
    subset=["transaction_hash"]
)

print(df.shape)
```

# SORT TEMPORAL

```python
df = df.sort_values(
    by="timestamp"
).reset_index(drop=True)
```

# DISTRIBUSI BULAN

```python
df["month"] = df["timestamp_dt"].dt.month

df["month"].value_counts().sort_index()
```

```python
output_path = "../data/processed/nft_clean.csv"

df.to_csv(output_path, index=False)

print("Saved:", output_path)
```
